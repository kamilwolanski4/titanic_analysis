#### Podsumowanie przeżywalności w zależności od płci

##### Analiza danych pokazuje, że płeć miała bardzo duży wpływ na przeżywalność pasażerów Titanica:
- Kobiety: ~72,7% pasażerek przeżyło
- Mężczyźni: ~19,1% pasażerów przeżyło

#### Podsumowanie przeżywalności w zależności od portu zaokrętowania

##### Analiza danych pokazuje, że port, z którego pasażerowie wsiedli na Titanica, również miał wpływ na ich szanse przeżycia:

- Cherbourg (C): ~55,6% pasażerów przeżyło
- Queenstown (Q): ~35,8% pasażerów przeżyło
- Southampton (S): ~33,3% pasażerów przeżyło

#### Przeżywalność pasażerów w zależności od klasy biletu i płci

##### Analiza danych wskazuje, że zarówno klasa biletu, jak i płeć miały istotny wpływ na przeżywalność pasażerów Titanica:

##### 1 klasa:
- Kobiety: ~96,5% przeżyło
- Mężczyźni: ~34,1% przeżyło

##### 2 klasa:
- Kobiety: ~88,7% przeżyło
- Mężczyźni: ~14,6% przeżyło

##### 3 klasa:
- Kobiety: ~49,1% przeżyło
- Mężczyźni: ~15,2% przeżyło

#### Kobiety miały znacznie większe szanse na przeżycie niż mężczyźni we wszystkich klasach.
#### Najwyższy odsetek przeżycia odnotowano wśród kobiet z 1 klasy (~96,5%).
#### Najniższy odsetek przeżycia dotyczył mężczyzn z 3 klasy (~15,2%).
#### Klasa biletu odgrywała kluczową rolę – im wyższa klasa, tym większe szanse na przeżycie
